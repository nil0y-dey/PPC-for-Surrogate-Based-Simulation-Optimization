## Setup

```python
import os, shutil, sys, subprocess

REPO_URL = "https://github.com/nil0y-dey/PPC-for-Surrogate-Based-Simulation-Optimization.git"
if "google.colab" in sys.modules:
    REPO_DIR = "/content/PPC-for-Surrogate-Based-Simulation-Optimization"
    if subprocess.run(["git", "-C", REPO_DIR, "pull", "-q"], capture_output=True).returncode:
        shutil.rmtree(REPO_DIR, ignore_errors=True)
        subprocess.run(["git", "clone", "-q", REPO_URL, REPO_DIR], check=True)
    from google.colab import drive
    drive.mount("/content/drive")
    BASE = "/content/drive/MyDrive/certified_stopping"
else:
    REPO_DIR = os.path.abspath("..") if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
    BASE = os.path.join(REPO_DIR, "colab_output")

os.chdir(REPO_DIR)
sys.path.insert(0, REPO_DIR)
OUT_DIR = os.path.join(BASE, "results")
PAPER_DIR = os.path.join(BASE, "paper")
os.makedirs(OUT_DIR, exist_ok=True)
os.environ["CS_CACHE_DIR"] = os.path.join(BASE, "cache")

from ppc.experiments import run_part, progress


def run_script(*args):
    result = subprocess.run([sys.executable, *args], capture_output=True, text=True)
    if result.returncode:
        raise RuntimeError(result.stderr)
```
